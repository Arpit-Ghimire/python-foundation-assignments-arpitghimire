# SQL Operators Assignment – Employee Dataset

## 1. Setup: Connect to PostgreSQL

```python
import psycopg2
import pandas as pd
import os
from dotenv import load_dotenv

#Load credentials from .env file
load_dotenv()

DB_CONFIG = {
    "host": os.getenv("DB_HOST", "localhost"),
    "port": os.getenv("DB_PORT", "5432"),
    "dbname": os.getenv("DB_NAME"),
    "user": os.getenv("DB_USER"),
    "password": os.getenv("DB_PASSWORD")
}

conn = psycopg2.connect(**DB_CONFIG)
conn.autocommit = True
cur = conn.cursor()
print("Connected to PostgreSQL successfully.")
```

## 2. Create the `employees` table

The table is created (or recreated, if it already exists) to match the 29 columns of the dataset.

```python
create_table_sql = """
DROP TABLE IF EXISTS employees;

CREATE TABLE employees (
    employee_id         INTEGER PRIMARY KEY,
    employee_code       VARCHAR(20),
    first_name          VARCHAR(50),
    last_name           VARCHAR(50),
    gender               VARCHAR(20),
    age                  INTEGER,
    city                 VARCHAR(50),
    department           VARCHAR(50),
    job_title            VARCHAR(100),
    employment_type      VARCHAR(30),
    join_date            DATE,
    years_experience     INTEGER,
    monthly_salary       NUMERIC(12,2),
    annual_bonus         NUMERIC(12,2),
    performance_rating   NUMERIC(3,1),
    projects_completed   INTEGER,
    leave_days_taken     INTEGER,
    overtime_hours       INTEGER,
    remote_worker        VARCHAR(5),
    employment_status    VARCHAR(20),
    education_level      VARCHAR(30),
    email                VARCHAR(100),
    phone                VARCHAR(20),
    emergency_contact    VARCHAR(20),
    manager_name         VARCHAR(100),
    certification        VARCHAR(50),
    work_shift           VARCHAR(20),
    performance_category VARCHAR(20),
    promotion_eligible   VARCHAR(5)
);
"""

cur.execute(create_table_sql)
print("Table 'employees' created.")
```

## 3. Load the dataset into the table

The dataset (`employees.csv`, exported from the original Excel file without modification)
is loaded row by row using an `INSERT` statement, with empty strings converted to `NULL`
so that `IS NULL` / `IS NOT NULL` questions work correctly.

```python
df=pd.read_excel("Employee_SQL_Operators_Assignment_29_Columns.xlsx",sheet_name="Dataset")
df.to_csv("employees.csv",index=False)
print("Converted successfully!")
```

```python

df = df.where(pd.notnull(df), None)   # convert NaN -> None (SQL NULL)

insert_sql = """
INSERT INTO employees (
    employee_id, employee_code, first_name, last_name, gender, age, city,
    department, job_title, employment_type, join_date, years_experience,
    monthly_salary, annual_bonus, performance_rating, projects_completed,
    leave_days_taken, overtime_hours, remote_worker, employment_status,
    education_level, email, phone, emergency_contact, manager_name,
    certification, work_shift, performance_category, promotion_eligible
) VALUES (
    %s, %s, %s, %s, %s, %s, %s,
    %s, %s, %s, %s, %s,
    %s, %s, %s, %s,
    %s, %s, %s, %s,
    %s, %s, %s, %s, %s,
    %s, %s, %s, %s
)
"""

rows = [tuple(row) for row in df.itertuples(index=False, name=None)]
cur.executemany(insert_sql, rows)

cur.execute("SELECT COUNT(*) FROM employees;")
print("Rows loaded:", cur.fetchone()[0])
```

## 4. Helper function

A small helper to run a query and return the result as a pandas DataFrame for easy viewing.

```python
def run_query(sql):
    """Run a SQL query and return results as a pandas DataFrame."""
    return pd.read_sql_query(sql, conn)
```
